```python
import pandas as pd 
df=pd.read_csv('creditcard.csv')
print(df.head())
```

```python
print("La taille de la dataset est:",df.shape)
print(df['Class'].value_counts())
```

```python
stats_normal = df[df['Class'] == 0].describe()

print("--- STATISTIQUES TRANSACTIONS NORMALES ---")
print(stats_normal)
```

```python
stats_fraude = df[df['Class'] == 1].describe()
print("\n--- STATISTIQUES FRAUDES ---")
print(stats_fraude)
```

```python
comparaison_moyennes = df.groupby('Class').mean()
print(comparaison_moyennes)
```

```python
df.drop_duplicates(inplace = True)
```
